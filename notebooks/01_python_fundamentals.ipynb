```python
# Use your existing classify_price() function to create a price_category column.

def classify_price(price):
    if price < 100:
        category = "Budget"
    elif price < 250:
        category = "Mid-Range"
    else: 
        category = "Premium"
    return category

df["price_category"] = df["price"].apply(classify_price)

df[["neighborhood", "price", "price_per_bedroom", "is_premium", "price_difference_from_mean", "price_category"]]
```
