## Denver Airbnb Analyis Notes - Exploratory Data Analysis

The Denver Airbnb listing snapshot is heavily concentrated in entire-home/apartment listings, which account for 4,233 of 4,939 listings (85.71%).

No individual neighborhood dominates Denver's Airbnb inventory. Inventory is geographically distributed across many neighborhoods, although there is still meaningful concentration among the leading neighborhoods.
- Top 5 neighborhoods → 26.67%
- Top 10 → 42.01%
- Top 13 → 49.32%
- Top 16 → 55.68%
    
Location Quotient Rank by Room Type:
- Overland's shared-room LQ of 41.01 is enormous: shared rooms are 43.18% of Overland's listings versus only 1.05% citywide. You also have 19 actual shared-room listings, so this isn't being driven by one or two observations.
- Berkeley's hotel-room LQ of 21.10 tells a different story. Hotel rooms are only 9.83% of Berkeley inventory, but they're just 0.47% of Denver inventory overall. Berkeley contains 17 of Denver's 23 hotel-room listings—73.91% of the city's total in this dataset.
- For private rooms, Gateway–Green Valley Ranch is particularly compelling because it combines concentration with substantial volume: 91 private rooms, 46.43% of neighborhood inventory, versus 12.78% citywide, giving LQ = 3.63.
- Entire homes behave differently. Auraria, Marston, Fort Logan, Country Club, Cory-Merrill, and University Park show LQ ≈ 1.17 because they're 100% entire-home listings in this dataset. But Denver itself is already 85.71% entire homes, so even a neighborhood at 100% cannot produce an extreme LQ:

Host Scale Across Denver Neighbourhoods:
- Neighborhoods were required to have at least 5 distinct hosts and 10 listings to reduce instability in host- and listing-share percentages caused by very small neighborhood samples. Large/professional operators are defined by having 10+ listings across the Denver market, rather than 10+ listings within an individual neighborhood.

Pricing Analysis:
- Denver nightly prices are strongly right-skewed, with a median of $176.45 versus a mean of $276.04, a small but meaningful high-price tail, and several exceptionally extreme observations.
- There are no low-price outliers under our room-type-specific IQR definition.
- Roughly 1 in 10 priced listings exceeds $500, but only 1.46% exceeds $1,000. Once we reach $2,000, we're down to only 9 listings, and above $5,000 there are only 4.
- Price anomaly: Marion — Cheesman Park: Entire condo, accommodates 2, 30-night minimum, $110,616.60/night, estimated occupancy 0.
- 235 total high-price outliers among Entire home/apartment and Private rooms
- There is a very clear price hierarchy by accommodation type. Hotel rooms have the highest median at $378, followed by Entire home/apartment at about $191, Private rooms at $106, and Shared rooms at only $35.
- For Entire home/apartment, the mean is about $107 above the median. That's substantial positive skew and consistent with the 213 high-price outliers we identified. Private rooms show the same pattern, although less dramatically: mean $152.98 versus median $106.
- For Entire home/apt, removing the IQR-defined high-price outliers reduces the mean from $297.35 to $217.78. That's a reduction of about 26.8%. Yet the median moves only from $190.59 to $180.62, about 5.2%.
- Private rooms tell essentially the same story: the mean falls by $40.67, or about 26.6%, while the median falls by only about 3.8%.
- Full dataset → describes the actual observed listing-price distribution.
- Median → primary measure of typical price.
- IQR price_status → allows us to identify/analyze unusual prices.
- Outlier-excluded mean → useful sensitivity/robustness measure.
- Neighborhood price comparisons are limited to neighborhoods with at least 20 listings containing a valid nightly price. This threshold retains 93.67% of priced listings while reducing the influence of neighborhood estimates based on small samples.
- Hampden deserves slightly more caution. It passes our threshold, but only with 24 priced listings, and its interquartile range is enormous: IQR = 572.64 - 99.77 = $472.87. So "$356.50 is Hampden's median" is supported by the data, but I wouldn't make a strong claim that Hampden is definitively Denver's most expensive Airbnb neighborhood without acknowledging its relatively small and highly dispersed sample. Cherry Creek, Jefferson Park, Highland, and Berkeley have progressively stronger sample support.
- Neighborhood-level Entire home/apt price comparisons require at least 15 priced Entire home/apt listings. This retains 55 neighborhoods and 95.1% of Denver's priced Entire home/apt inventory while excluding neighborhoods with very small comparison samples.
- City Park, Capitol Hill, and Overland remain among the least expensive neighborhoods even after restricting the analysis to Entire homes/apartments.
- On the expensive side, Cherry Creek and Jefferson Park remain expensive after controlling for room type, while Hampden drops relative to them. This gives us evidence that neighborhood itself—or characteristics associated with those neighborhoods—is related to price beyond just room-type composition.
- Among priced Entire home/apt listings with 1–3 bedrooms, single-listing hosts have higher median nightly prices than the Denver-wide median for each corresponding bedroom count, while small and medium multi-listing operators consistently fall below those benchmarks. Large/professional operators show a mixed pattern, with below-benchmark 1–2 bedroom prices but above-benchmark 3-bedroom prices.
- Among the 41 comparable neighborhood × bedroom segments with at least three listings from each host type, Single-listing hosts have a higher median nightly price in 28 segments (68.3%), while large/professional operators have a higher median in 13 segments (31.7%).
- After controlling for neighborhood and bedroom count among comparable 1–3 bedroom Entire home/apt segments, Single-listing hosts had higher median nightly prices in 68.3% of comparisons. Across all 41 comparisons, the median signed difference was +28.5% relative to large/professional operators, while the typical absolute pricing gap was 35.6%.

Minimum night strategy analysis 
- Longer minimum-stay policies are associated with substantially lower nightly prices. The relationship persists after controlling for room type, particularly among entire-home/apartment listings, where monthly-stay listings have a median nightly price approximately 60% below short-stay listings.
- Among comparable 1–3 bedroom entire homes/apartments, monthly-stay listings have median nightly prices approximately 54–59% lower than short-stay listings. The discount persists across all three bedroom counts, suggesting that the relationship between longer minimum-stay requirements and lower nightly rates is not simply explained by differences in property size.
- Monthly-stay Entire home/apt listings have median nightly rates approximately 54–59% lower than comparable short-stay listings across 1-, 2-, and 3-bedroom properties.

Amenity analysis
- Amenity richness remains positively associated with nightly price even after controlling for room type and bedroom count, particularly among 2- and 3-bedroom entire-home listings. The relationship is weaker among 1-bedroom listings.
- Hot tubs are associated with a substantial nightly price premium among Denver entire-home Airbnb listings. After controlling for bedroom count, listings with hot tubs have median nightly prices approximately 16% higher for one-bedroom properties and 49% higher for both two- and three-bedroom properties.
- Amenity quantity: Listings with richer amenity sets tend to have higher nightly prices, particularly among larger properties.
- Amenity composition: Hot tubs specifically are associated with higher median prices: +16% for 1-bedroom and approximately +49% for 2- and 3-bedroom entire-home listings.
- Pools do not show the same pattern

Estimated occupancy analysis 
- Entire homes/apartments have substantially higher estimated occupancy than the other room types, and within entire homes/apartments, smaller 1-bedroom properties have higher estimated occupancy than 2- and 3-bedroom properties.
- Among comparable 1–3 bedroom entire-home listings, monthly-stay properties have substantially lower modeled occupancy than short-stay properties. Approximately 41–54% of monthly-stay listings have zero estimated occupied nights, compared with only 3–6% of short-stay listings.
- Estimated occupancy varies substantially across Denver neighborhoods even after restricting comparisons to similar listing types and bedroom counts, suggesting geography is associated with performance beyond differences in broad property composition.
- Listing structure matters: entire homes/apartments have substantially higher estimated occupancy than the other broad room types. Stay strategy matters: short-stay listings dramatically outperform monthly-stay listings in estimated occupancy, and monthly listings have extraordinarily high zero-estimated-occupancy rates. Location matters: substantial neighborhood differences remain even when controlling for room type, stay strategy, and bedroom count.
- After controlling for broad listing type, stay policy, and bedroom count, the relationship between nightly price and estimated occupancy does not appear strongly linear from the grouped medians alone.
- Among comparable 1–3 bedroom short-stay entire homes/apartments, Superhost listings consistently show substantially higher estimated occupancy than non-Superhost listings. Median estimated occupancy is roughly 92–119 nights higher for Superhosts depending on bedroom count, while zero-estimated-occupancy rates are substantially lower.
- Roughly 81% of the original 5.0 listings disappear when we require 20 reviews.
- Guest rating alone showed a misleading relationship with estimated occupancy. Perfect 5.0 listings initially had much lower estimated occupancy than listings rated 4.5–4.99, but perfect-rated listings also had substantially fewer reviews. After restricting the comparison to listings with at least 20 reviews, the median estimated occupancy gap narrowed from 80 nights to 14 nights (168 vs. 154), suggesting listing maturity/review volume explains much of the original difference.
- Single-listing and medium-scale hosts show substantially higher estimated occupancy per listing than large/professional operators, while small multi-listing hosts fall between those extremes.
- Among comparable 1–3 bedroom short-stay entire homes/apartments, single-listing hosts had substantially higher estimated occupancy than large/professional operators, with median estimated occupancy of 161 versus 84 nights. Large/professional operators had a somewhat higher incidence of zero estimated occupancy (6.8% vs. 3.87%), but this difference is too small to account for the overall occupancy gap, suggesting lower estimated occupancy is more broadly distributed across their listings. Medium operators were an exception, showing occupancy comparable to single-listing hosts, although the group contained only 81 listings.

Estimated revenue analysis
- Among comparable short-stay entire homes/apartments, larger listings commanded substantially higher nightly prices but had lower estimated occupancy. Median estimated annual revenue nevertheless increased from approximately $33.0K for one-bedroom listings to $44.1K for two-bedroom listings and $47.5K for three-bedroom listings, suggesting higher nightly rates offset lower utilization. However, the revenue gain from two to three bedrooms was relatively modest despite a substantial increase in nightly price.
- The highest-priced neighborhoods are not necessarily the highest-revenue neighborhoods. Estimated revenue appears to depend on the interaction between nightly price and estimated occupancy.
- Revenue performance appears strongest when listings achieve a favorable balance between nightly price and utilization, rather than maximizing either one independently.
- Large/professional operators tend to command relatively high nightly prices without achieving correspondingly high estimated revenue. Their estimated occupancy is lower than single-listing and medium operators at every bedroom count.
- Among comparable short-stay entire homes/apartments, Superhost listings are associated with substantially higher estimated annual revenue across 1-, 2-, and 3-bedroom properties. The difference appears to be driven primarily by much higher estimated occupancy, with additional nightly-price advantages among 2- and 3-bedroom listings.
- Host portfolio size doesn't predict success cleanly, but Superhost status does show a strong association with estimated performance.
- Within established, highly rated listings, perfect 5.0 ratings are associated with higher estimated revenue, but not necessarily higher estimated occupancy.
- Among established, highly rated comparable listings, perfect 5.0-rated properties are associated with higher median estimated annual revenue across all three bedroom categories. The results suggest a price premium is part of this relationship, while 5.0 ratings do not consistently correspond to higher estimated occupancy.
- The 5.0 groups are much smaller and have substantially fewer reviews. For example, 1BR 5.0 listings have a median of 46 reviews versus 137 for 4.5–4.99 listings. So 5.0 shouldn't automatically be interpreted as superior operating performance.
- Among comparable short-stay entire homes/apartments, greater amenity richness is generally associated with higher estimated annual revenue. The pattern persists within bedroom categories and appears to reflect a combination of stronger nightly pricing and higher estimated utilization, particularly among listings with roughly 40–70+ amenities.
- Higher nightly prices are not uniformly associated with higher estimated annual revenue. Among comparable Denver short-stay listings, increasing price is generally accompanied by lower estimated occupancy, and several bedroom segments show a middle-to-upper price range where the combination of rate and utilization produces higher estimated revenue than still-more-expensive listings.
- Bedroom count affects pricing, price relates to occupancy, occupancy strongly affects revenue, and therefore revenue performance reflects the interaction between rate and utilization rather than either metric alone.

Host location category analysis
- Out-of-state hosts have the highest median estimated revenue. Their median estimated revenue is $44,709, compared with $39,652 for Denver-based hosts. However, they also have higher median nightly prices ($349.50 versus $267.33) and lower estimated occupancy (133 versus 161 nights). This suggests that out-of-state hosts may be following a higher-price, lower-utilization strategy.
- Denver-based hosts have the highest median estimated occupancy. At 161 nights, Denver-based hosts outperform the other visible categories in estimated occupancy. They also represent by far the largest segment, with 1,297 listings. This is an interesting finding because local hosts appear to achieve greater utilization without charging the highest nightly rates.
- Other Colorado hosts have the lowest median estimated revenue. Their median revenue is $23,756.50, substantially below both Denver and out-of-state hosts. But this does not necessarily mean these hosts are less effective. Differences in bedroom distribution, neighborhoods, property characteristics, and listing maturity could explain some of the gap.

Property category analysis 
- Property type matters even after controlling for bedroom count and stay policy. Median estimated revenue differs considerably between property categories.
- Price and occupancy can offset one another. Guesthouses generally have stronger estimated occupancy, whereas condos command higher nightly prices.
- Sample size matters. The 294 one-bedroom guesthouses provide a more substantial comparison group than the 11 three-bedroom condos.

Bathrooms analysis 
- Additional bathrooms are associated with a pricing premium within most bedroom categories.
- Higher nightly prices do not necessarily translate into proportionally higher estimated revenue, because estimated occupancy can be lower.
- The strongest revenue advantage appears among two-bedroom listings, where higher bathroom counts are associated with progressively higher median estimated revenue.

listing acitivty status analysis 
- Recently reviewed listings dominate estimated performance
- Recently reviewed listings have: The highest median nightly price ($236.78). The highest median estimated occupancy (175 nights). The highest median estimated annual revenue ($36,726).
- Their median estimated revenue is approximately 2.9 times that of listings reviewed within the last year but not within the last 30 days.
- This is a substantial difference, although it shouldn't automatically be interpreted as evidence that recent reviews cause higher occupancy.
- Zero estimated occupancy aligns perfectly with two activity categories.
- Every listing in the historically reviewed and never-reviewed categories has zero estimated occupancy in your filtered results.
- Conversely, every listing with a review within the last year has positive estimated occupancy.
- This is too systematic to treat as an ordinary market correlation.
- It strongly suggests that estimated_occupancy_l365d is derived using recent review activity, or at least depends on a related underlying measure.
- We haven't independently verified the dataset provider's methodology, so we shouldn't claim to know the exact formula yet.

hosts_time_as_host_years analysis
- Finding A: Hosts with 1–5 years of experience have the highest median estimated revenue across all three bedroom groups. The differences are substantial. However, this does not establish that gaining experience directly increases revenue. Newer hosts may have incomplete operating histories, different listing characteristics, or different review activity.
- Finding B: New hosts have the highest zero-occupancy percentages. The pattern is particularly pronounced for 3-bedroom listings, where 14.29% of listings hosted by people with less than one year of experience have zero modeled occupancy. The sample for this group is only 28 listings, so the percentage should be interpreted cautiously.
- Finding C: The relationship is not consistently linear beyond the first year. For 1-bedroom listings, hosts with more than 10 years of experience have higher median estimated occupancy than hosts with 1–5 years: Over 10 years: 203 days. 1–5 years: 196 days.
- However, their median estimated revenue is lower. This illustrates why occupancy and revenue need to be evaluated together. Higher modeled occupancy doesn't necessarily translate into higher modeled revenue when pricing differs. For 2- and 3-bedroom listings, the 1–5 year category has the highest median estimated revenue, but the relationships between tenure, price, and occupancy vary.
- Among comparable short-stay entire homes/apartments, hosts with 1–5 years of tenure have the highest median modeled annual revenue across all three bedroom groups. The association does not establish a causal benefit of host experience.

reviews_per_month_analysis
- Review frequency and estimated occupancy are strongly associated across several categories, but the occupancy metric may depend on review information.
- Estimated occupancy appears to have a ceiling of 255 days, which limits comparisons among higher-activity listings.

review_scores_location analysis

Even after excluding listings with fewer than 20 reviews, the highest-ranked neighborhoods have average ratings between approximately 4.92 and 4.98. This supports our earlier observation that location ratings are highly concentrated near the maximum of 5.0. The difference between Washington Park (4.98) and Highland (4.92) is just 0.06 points on a five-point scale. That may be statistically detectable in some circumstances, but it is not necessarily practically meaningful.

Likewise, Highland has 105 qualifying listings, giving us a broader representation of qualifying properties than neighborhoods with only 10–11. This doesn't automatically make the larger groups unbiased, but it makes their descriptive summaries less dependent on a handful of properties.

This is consistent with ratings being tightly concentrated toward the upper end of the scale. However, mean and median alone cannot establish the full shape of the distribution. A boxplot in Python would make the spread easier to examine.

cleaned_calendar analysis
- Availability is lowest near the beginning of the calendar. July 2026 has an availability percentage of 31.26%, compared with approximately 62–67% in November and December 2026. This could reflect bookings, blocked dates, or the fact that dates closer to the scrape are more likely to have already been reserved. We cannot interpret this as proof that July has the highest demand.
- Availability rises later in 2026 November and December have the highest availability percentages in your results: November: 66.14% December: 66.55%
- This means approximately two-thirds of the observed listing-days were marked available. It does not mean these months had a 33–34% occupancy rate.
- Availability declines again in spring 2027. April, May, and June 2027 have availability percentages near 51%. That could reflect seasonal patterns, advance booking restrictions, or differences in how hosts manage future availability. We would need additional evidence to distinguish these explanations.
- Monthly-stay listings generally have higher future availability. From September 2026 through June 2027, monthly-stay listings consistently have a higher percentage of available listing-days than short-stay listings. The gap becomes particularly noticeable in spring 2027.
- Short-stay availability declines more sharply in spring 2027 Between March and April 2027: Short-stay availability falls from 58.36% to 45.57%. Monthly-stay availability falls from 64.91% to 56.08%.
- This suggests different future calendar availability patterns between the two stay policies. It does not necessarily mean short-stay properties will achieve higher occupancy. Unavailable dates can represent reservations, owner blocks, or other restrictions.
- The pattern reverses near the beginning of the calendar. In July and August 2026, short-stay listings actually have higher availability percentages.
- This is worth noting because the relationship between stay policy and availability isn't constant across the entire booking horizon.
- Business interpretation: Stay policy appears to be associated with differences in how hosts manage future availability, particularly further into the calendar. We cannot attribute the differences to demand alone.
- Availability differs substantially by neighborhood. Hale has 78.14% of its future listing-days available, compared with only 35.76% in Washington Park. That's a difference of 42.38 percentage points. This suggests that neighborhood is associated with meaningful differences in future calendar availability, even after restricting the analysis to comparable property types.
- High estimated revenue doesn't necessarily correspond to low availability. Earlier, we identified Five Points as a relatively strong neighborhood for estimated revenue.
- In this analysis, Five Points has: 143 qualifying listings, 53.22% future availability. Meanwhile, Highland has 133 listings and 62.92% availability.
- Both neighborhoods appeared competitive in our earlier revenue analysis, yet their future availability profiles differ.
- This reinforces an important distinction: Estimated annual revenue measures modeled historical performance, whereas calendar availability describes future bookable inventory. We should not interpret the unavailable percentage as an occupancy rate because blocked dates and booked dates are not distinguishable from this field alone.

cleaned_reviews analysis
- Review activity shows consistent seasonality. The strongest review activity generally occurs between June and September, while January, February, and December have substantially fewer reviews.
- 2023 peak: July, 5,873 reviews.
- 2024 peak: June, 6,532 reviews.
- 2025 peak: June, 7,009 reviews.
- This recurring pattern is stronger evidence of seasonality than examining any single year.
- Review volume increased across the three years. Nearly every month shows higher review volume in 2025 than in 2023.
- However, this does not necessarily mean Airbnb demand increased by the same proportion. The number of listings represented in the historical review data may have changed, and reviews are only an indirect indicator of guest activity.
- Reviews per reviewed listing also fluctuate seasonally For example, in 2023:
- January: 2.86 reviews per reviewed listing.
- July: 3.97.
- December: 2.44.
- This suggests the seasonal pattern is not explained solely by changes in the number of listings receiving reviews. Review frequency among those listings also varies.

## Does reviews per month have an effect on median price, median estimated occupancy, and median estimated revenue?

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified, hosts_time_as_host_years,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Specialty/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- Does reviews per month have an effect on median price, median estimated occupancy, and median estimated revenue?


SELECT 
	CASE 
		WHEN reviews_per_month IS NULL THEN 'Missing'
		WHEN reviews_per_month < 1.00 THEN 'Under 1.00'
		WHEN reviews_per_month < 4.00 THEN '1.00-3.99'
		WHEN reviews_per_month < 7.00 THEN '4.00-6.99'
		WHEN reviews_per_month < 10.00 THEN '7.00-9.99'
		ELSE '10.00+'
	END AS reviews_per_month_category,
	COUNT(*) AS listings,
	ROUND(AVG(review_scores_rating), 2) AS avg_rating_score,
	MEDIAN(cleaned_price) AS median_price,
	MEDIAN(estimated_occupancy_l365d) AS median_estimated_occupancy,
	SUM(CASE WHEN estimated_occupancy_l365d = 0 THEN 1 ELSE 0 END) AS zero_estimated_occupancy_values,
	ROUND(
    	COUNT(*) FILTER (
        WHERE estimated_occupancy_l365d = 255
    	) * 100.0 / COUNT(*),
    2) AS pct_at_occupancy_ceiling,
	MEDIAN(estimated_revenue_l365d) AS median_estimated_revenue
FROM cleaned_listings
WHERE cleaned_price IS NOT NULL AND estimated_revenue_l365d IS NOT NULL 
	AND room_type = 'Entire home/apt' AND stay_policy = 'Short stay' AND bedrooms BETWEEN 1 AND 3
GROUP BY reviews_per_month_category
ORDER BY listings DESC;
```

## Does avg_review_scores_location have significant differences by neighborhood?

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified, hosts_time_as_host_years,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Specialty/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- How much does avg_review_scores_location differ by neighborhood?
	
SELECT 
	neighbourhood_cleansed,
	COUNT(*) AS listings,
	ROUND(AVG(number_of_reviews), 2) AS avg_reviews_per_qualified_listing,
	ROUND(AVG(review_scores_location), 2) AS avg_review_scores_location,
	MEDIAN(review_scores_location) AS median_review_scores_location
FROM cleaned_listings
WHERE room_type = 'Entire home/apt' AND stay_policy = 'Short stay' AND bedrooms BETWEEN 1 AND 3
	AND number_of_reviews >= 20 
GROUP BY neighbourhood_cleansed
HAVING COUNT(*) >= 10
ORDER BY avg_review_scores_location DESC, listings DESC;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified, hosts_time_as_host_years,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Specialty/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- Does time as host (years) impact median price, median estimated occupancy, or median estimated revenue?
	
SELECT 
	CASE 
		WHEN hosts_time_as_host_years < 1 THEN 'Less Than 1 Year'
		WHEN hosts_time_as_host_years BETWEEN 1 AND 5 THEN '1-5 Years'
		WHEN hosts_time_as_host_years BETWEEN 6 AND 10 THEN '6-10 Years'
		ELSE 'Over 10 Years'
	END AS category_hosts_time_as_host_years,
	bedrooms,
	COUNT(*) AS listings,
	MEDIAN(cleaned_price) AS median_price,
	MEDIAN(estimated_revenue_l365d) AS median_estimated_revenue,
	MEDIAN(estimated_occupancy_l365d) AS median_estimated_occupancy,
	SUM(CASE WHEN estimated_occupancy_l365d = 0 THEN 1 ELSE 0 END) AS zero_estimated_occupancy_listing,
	ROUND(
    SUM(CASE WHEN estimated_occupancy_l365d = 0 THEN 1 ELSE 0 END)
    * 100.0 / COUNT(*),
    2
) AS pct_zero_estimated_occupancy
FROM cleaned_listings
WHERE cleaned_price IS NOT NULL and hosts_time_as_host_years IS NOT NULL AND estimated_revenue_l365d IS NOT NULL
	AND room_type = 'Entire home/apt' AND stay_policy = 'Short stay' AND bedrooms BETWEEN 1 AND 3
GROUP BY category_hosts_time_as_host_years, bedrooms
ORDER BY bedrooms, category_hosts_time_as_host_years;
```

## Which Denver neighborhoods have the highest and lowest future calendar availability among comparable short-stay entire homes/apartments?

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Specialty/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month,
		l.neighbourhood_cleansed,
		l.room_type,
		l.bedrooms,
		l.stay_policy
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- Which Denver neighborhoods have the highest and lowest future calendar availability among comparable short-stay entire homes/apartments?

SELECT 
	neighbourhood_cleansed,
	COUNT(DISTINCT listing_id) AS listings,
	COUNT(*) AS listing_days,
	SUM(CASE WHEN available IS TRUE THEN 1 ELSE 0 END) AS available_listing_days,
	SUM(CASE WHEN available IS FALSE THEN 1 ELSE 0 END) AS unavailable_listing_days,
	ROUND(SUM(CASE WHEN available IS TRUE THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS pct_of_listing_days_available
FROM cleaned_calendar
WHERE year_month >= DATE '2026-07-01'
	AND year_month < DATE '2027-07-01'
	AND available IS NOT NULL
	AND date IS NOT NULL 
	AND room_type = 'Entire home/apt' AND stay_policy = 'Short stay' AND bedrooms BETWEEN 1 AND 3
GROUP BY neighbourhood_cleansed
HAVING COUNT(DISTINCT listing_id) >= 20
ORDER BY pct_of_listing_days_available DESC;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Specialty/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- How does listing activity status relate to median nightly price, median estimated occupancy, and median estimated annual revenue among Denver Airbnb listings?

SELECT
	listing_activity_status,
	COUNT(*) AS listings,
	MEDIAN(cleaned_price) AS median_price,
	MEDIAN(estimated_occupancy_l365d) AS median_estimated_occupancy,
	MEDIAN(estimated_revenue_l365d) AS median_estimated_revenue,
	SUM(CASE WHEN estimated_occupancy_l365d = 0 THEN 1 ELSE 0 END) AS zero_occupancy_listings
FROM cleaned_listings
WHERE estimated_revenue_l365d IS NOT NULL AND cleaned_price IS NOT NULL
GROUP BY listing_activity_status
ORDER BY median_estimated_revenue DESC;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Specialty/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- How do different Denver Airbnb property categories compare in listing count, median nightly price, median estimated occupancy, and median estimated annual revenue?
-- Among comparable 1–3 bedroom short-stay entire homes/apartments, how do property categories differ in median nightly price, median estimated occupancy, and median estimated annual revenue?

SELECT 
	property_category,
	bedrooms,
	COUNT(*) AS listings,
	COUNT(DISTINCT host_id) AS hosts,
	ROUND(COUNT(*) / COUNT(DISTINCT host_id), 2) AS listings_per_host,
	MEDIAN(cleaned_price) AS median_price,
	MEDIAN(estimated_occupancy_l365d) AS median_estimated_occupancy,
	MEDIAN(estimated_revenue_l365d) AS median_estimated_revenue
FROM cleaned_listings
WHERE estimated_revenue_l365d IS NOT NULL AND cleaned_price IS NOT NULL 
	AND room_type = 'Entire home/apt' AND stay_policy = 'Short stay' AND bedrooms BETWEEN 1 AND 3
GROUP BY property_category, bedrooms
HAVING COUNT(*) >= 10
ORDER BY bedrooms, median_estimated_revenue DESC;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

SELECT 
	host_location_category,
	COUNT(DISTINCT host_id) AS hosts,
	COUNT(*) AS listings,
	SUM(CASE WHEN host_scale = 'Single-listing host' THEN 1 ELSE 0 END) AS single_host_listings,
	SUM(CASE WHEN host_scale = 'Small multi-listing host' THEN 1 ELSE 0 END) AS small_multi_host_listings,
	SUM(CASE WHEN host_scale = 'Medium operator' THEN 1 ELSE 0 END) AS medium_operator_listings,
	SUM(CASE WHEN host_scale = 'Large/professional operator' THEN 1 ELSE 0 END) AS large_professional_operator_listings
FROM cleaned_listings
WHERE estimated_revenue_l365d IS NOT NULL AND cleaned_price IS NOT NULL
GROUP BY host_location_category
ORDER BY listings DESC;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- How do host location categories differ in listing count, median nightly price, median estimated occupancy, and median estimated annual revenue?

SELECT 
	host_location_category,
	COUNT(DISTINCT host_id) AS hosts,
	COUNT(*) AS listings,
	MEDIAN(cleaned_price) AS median_price,
	MEDIAN(estimated_occupancy_l365d) AS median_estimated_occupancy,
	MEDIAN(estimated_revenue_l365d) AS median_estimated_revenue
FROM cleaned_listings
WHERE estimated_revenue_l365d IS NOT NULL AND cleaned_price IS NOT NULL 
	AND room_type = 'Entire home/apt' AND stay_policy = 'Short stay' AND bedrooms BETWEEN 1 AND 3
GROUP BY host_location_category
ORDER BY median_estimated_revenue DESC;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- Among comparable 1–3 bedroom short-stay entire homes/apartments, how does nightly price relate to estimated annual revenue?

SELECT 
	bedrooms,
	CASE 
		WHEN cleaned_price < 100 THEN '< $100'
		WHEN cleaned_price < 200 THEN '$100-199'
		WHEN cleaned_price < 300 THEN '$200-299'
		WHEN cleaned_price < 400 THEN '$300-399'
		WHEN cleaned_price < 500 THEN '$400-499'
		WHEN cleaned_price < 750 THEN '$500-749'
		WHEN cleaned_price < 1000 THEN '$749-999'
	ELSE '$1,000+'
	END AS price_category,
	COUNT(*) AS listings,
	MEDIAN(cleaned_price) AS median_price,
	MEDIAN(estimated_occupancy_l365d) AS median_estimated_occupancy,
	MEDIAN(estimated_revenue_l365d) AS median_estimated_revenue
FROM cleaned_listings
WHERE estimated_revenue_l365d IS NOT NULL AND cleaned_price IS NOT NULL 
	AND room_type = 'Entire home/apt' AND stay_policy = 'Short stay' AND bedrooms BETWEEN 1 AND 3
GROUP BY bedrooms, price_category
HAVING COUNT(*) >= 10
ORDER BY bedrooms, median_estimated_revenue DESC;
```

## Among comparable 1–3 bedroom short-stay entire homes/apartments, how does Superhost status relate to estimated annual revenue?

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- Among comparable 1–3 bedroom short-stay entire homes/apartments, how does Superhost status relate to estimated annual revenue?

SELECT 
	host_is_superhost,
	bedrooms,
	COUNT(*) AS listings,
	MEDIAN(cleaned_price) AS median_price,
	MEDIAN(estimated_occupancy_l365d) AS median_estimated_occupancy,
	MEDIAN(estimated_revenue_l365d) AS median_estimated_revenue
FROM cleaned_listings
WHERE estimated_revenue_l365d IS NOT NULL AND cleaned_price IS NOT NULL 
	AND room_type = 'Entire home/apt' AND stay_policy = 'Short stay' AND bedrooms BETWEEN 1 AND 3
GROUP BY host_is_superhost, bedrooms
ORDER BY bedrooms, median_estimated_revenue DESC;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- Among comparable 1–3 bedroom short-stay entire homes/apartments, how does host scale relate to estimated annual revenue
-- Within comparable bedroom counts, do differences in estimated revenue by host scale persist?

host_scale_revenue_analysis AS (
	SELECT 
		host_scale,
		bedrooms,
		COUNT(*) AS listings,
		MEDIAN(cleaned_price) AS median_price,
		MEDIAN(estimated_occupancy_l365d) AS median_estimated_occupancy,
		MEDIAN(estimated_revenue_l365d) AS median_estimated_revenue
	FROM cleaned_listings
	WHERE estimated_revenue_l365d IS NOT NULL AND cleaned_price IS NOT NULL 
		AND room_type = 'Entire home/apt' AND stay_policy = 'Short stay' AND bedrooms BETWEEN 1 AND 3
	GROUP BY host_scale, bedrooms
)

SELECT 
	bedrooms,
	MAX(CASE WHEN host_scale = 'Single-listing host' THEN median_estimated_revenue END) AS single_median_estimated_revenue,
	MAX(CASE WHEN host_scale = 'Small multi-listing host' THEN median_estimated_revenue END) AS small_multi_median_estimated_revenue,
	MAX(CASE WHEN host_scale = 'Medium operator' THEN median_estimated_revenue END) AS medium_median_estimated_revenue,
	MAX(CASE WHEN host_scale = 'Large/professional operator' THEN median_estimated_revenue END) AS large_pro_median_estimated_revenue
FROM host_scale_revenue_analysis
GROUP BY bedrooms
ORDER BY bedrooms;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- Among comparable 1–3 bedroom short-stay entire homes/apartments, how does estimated annual revenue differ by bedroom count?

SELECT 
	bedrooms,
	COUNT(*) AS listings,
	MEDIAN(cleaned_price) AS median_price,
	MEDIAN(estimated_occupancy_l365d) AS median_estimated_occupancy,
	MEDIAN(estimated_revenue_l365d) AS median_estimated_revenue
FROM cleaned_listings
WHERE estimated_revenue_l365d IS NOT NULL AND cleaned_price IS NOT NULL 
	AND room_type = 'Entire home/apt' AND stay_policy = 'Short stay' AND bedrooms BETWEEN 1 AND 3
GROUP BY bedrooms
ORDER BY bedrooms;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- How does estimated annual revenue differ across the major types of Denver Airbnb listings?

SELECT 
	CASE 
		WHEN estimated_revenue_l365d < 1000 THEN '< $1,000'
		WHEN estimated_revenue_l365d < 10000 THEN '$1,000 - $9,999'
		WHEN estimated_revenue_l365d < 40000 THEN '$10,000 - $39,999'
		WHEN estimated_revenue_l365d < 70000 THEN '$40,000 - $69,999'
		WHEN estimated_revenue_l365d < 100000 THEN '$70,000 - $99,999'
		WHEN estimated_revenue_l365d < 200000 THEN '$100,000 - $199,999'
		ELSE '$200,000+'
	END AS estimated_revenue_category,
	COUNT(*) AS listings,
	MEDIAN(cleaned_price) AS median_price,
	ROUND(AVG(cleaned_price), 2) AS avg_price,
	MEDIAN(estimated_occupancy_l365d) AS median_estimated_occupancy,
	ROUND(AVG(estimated_occupancy_l365d), 2) AS avg_estimated_occupancy,
	SUM(CASE WHEN estimated_occupancy_l365d = 0 THEN 1 ELSE 0 END) AS zero_occupancy_listings
FROM cleaned_listings
WHERE estimated_revenue_l365d IS NOT NULL AND cleaned_price IS NOT NULL
GROUP BY estimated_revenue_category
ORDER BY median_price DESC;
```

## How does host scale relate to estimated occupancy among comparable short-stay entire homes/apartments?

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- How does host scale relate to estimated occupancy among comparable short-stay entire homes/apartments?

SELECT 
	host_scale,
	COUNT(*) AS listings,
	MEDIAN(estimated_occupancy_l365d) AS median_estimated_occupancy,
	ROUND(AVG(estimated_occupancy_l365d), 2) AS avg_estimated_occupancy,
	SUM(CASE
    WHEN estimated_occupancy_l365d = 0 THEN 1
    ELSE 0
	END) AS zero_occupancy_listings,
	ROUND(
    	SUM(CASE
       		WHEN estimated_occupancy_l365d = 0 THEN 1
        	ELSE 0
    	END) * 100.0 / COUNT(*),
    2) AS pct_zero_occupancy
FROM cleaned_listings
WHERE room_type = 'Entire home/apt' AND stay_policy = 'Short stay' AND bedrooms BETWEEN 1 AND 3
GROUP BY host_scale
ORDER BY median_estimated_occupancy DESC;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- How do guest ratings relate to estimated occupancy?
-- Do perfect-5.0 listings have fewer reviews than listings rated 4.5–4.99?
-- Among listings with comparable review histories, do higher-rated listings tend to have higher estimated occupancy?

SELECT 
	CASE 
		WHEN review_scores_rating < 4 THEN '<4.0'
		WHEN review_scores_rating <= 4.49 THEN '4.0-4.49'
		WHEN review_scores_rating <= 4.99 THEN '4.5-4.99'
		ELSE '5.0'
	END AS rating_category,
	COUNT(*) AS listings,
	MEDIAN(number_of_reviews) AS median_number_of_reviews,
	MEDIAN(estimated_occupancy_l365d) AS median_estimated_occupancy,
	ROUND(AVG(estimated_occupancy_l365d), 2) AS avg_estimated_occupancy
FROM cleaned_listings
WHERE review_scores_rating IS NOT NULL AND number_of_reviews >= 20
GROUP BY rating_category
ORDER BY median_estimated_occupancy DESC;
```

## Do Superhosts achieve higher estimated occupancy than non-Superhosts for comparable listings?

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- Do Superhosts achieve higher estimated occupancy than non-Superhosts for comparable listings?
-- Within comparable 1–3 bedroom short-stay entire homes/apartments, how does estimated occupancy differ between Superhosts and non-Superhosts?

SELECT 
	host_is_superhost,
	bedrooms,
	COUNT(*) AS listings,
	MEDIAN(estimated_occupancy_l365d) AS median_estimated_occupancy,
	ROUND(AVG(estimated_occupancy_l365d), 2) AS avg_estimated_occupancy,
	SUM(CASE WHEN estimated_occupancy_l365d = 0 THEN 1 ELSE 0 END) AS zero_occupancy_listings,
	ROUND(SUM(CASE WHEN estimated_occupancy_l365d = 0 THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS zero_occupancy_percentage
FROM cleaned_listings
WHERE room_type = 'Entire home/apt' AND stay_policy = 'Short stay' AND bedrooms BETWEEN 1 AND 3
GROUP BY host_is_superhost, bedrooms
ORDER BY host_is_superhost, bedrooms;
```

## How does nightly price relate to estimated occupancy?

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- How does nightly price relate to estimated occupancy?
-- Within comparable short-stay entire homes/apartments of the same bedroom size, how does nightly price vary across estimated occupancy levels?

SELECT 
	CASE 
		WHEN estimated_occupancy_l365d = 0 THEN '0 Days'
		WHEN estimated_occupancy_l365d <= 50 THEN '1-50 Days'
		WHEN estimated_occupancy_l365d <= 100 THEN '51-100 Days'
		WHEN estimated_occupancy_l365d <= 150 THEN '101-150 Days'
		WHEN estimated_occupancy_l365d <= 200 THEN '151-200 Days'
		ELSE 'Over 200 Days'
	END AS estimated_occupancy_category,
	bedrooms,
	COUNT(*) AS listings,
	MEDIAN(cleaned_price) AS median_price
FROM cleaned_listings
WHERE cleaned_price IS NOT NULL AND room_type = 'Entire home/apt' AND stay_policy = 'Short stay' AND bedrooms BETWEEN 1 AND 3
GROUP by estimated_occupancy_category, bedrooms
ORDER BY estimated_occupancy_category, bedrooms;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- Among comparable short-stay entire homes/apartments, which Denver neighborhoods have the highest and lowest estimated occupancy?

SELECT 
	neighbourhood_cleansed,
	bedrooms,
	COUNT(*) AS listings,
	SUM(COUNT(*)) OVER (PARTITION BY neighbourhood_cleansed) AS total_comparable_neighbourhood_listings,
 	MEDIAN(estimated_occupancy_l365d) AS median_estimated_occupancy,
	ROUND(AVG(estimated_occupancy_l365d), 2) AS avg_estimated_occupancy,
	SUM(CASE WHEN estimated_occupancy_l365d = 0 THEN 1 ELSE 0 END) AS zero_estimated_occupancy_listings,
	ROUND(SUM(CASE WHEN estimated_occupancy_l365d = 0 THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS pct_zero_occupancy
FROM cleaned_listings 
WHERE room_type = 'Entire home/apt'
  AND stay_policy = 'Short stay'
  AND bedrooms BETWEEN 1 AND 3
GROUP BY neighbourhood_cleansed, bedrooms
HAVING COUNT(*) >= 20
ORDER BY bedrooms, avg_estimated_occ;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- For each stay policy and bedroom count, what percentage of listings have estimated_occupancy_l365d = 0?

stay_policy_bedrooms_occupancy AS (
	SELECT 
		stay_policy,
		bedrooms,
		COUNT(*) AS listings,
		SUM(CASE WHEN estimated_occupancy_l365d = 0 THEN 1 ELSE 0 END) AS zero_occupancy_listings,
		MEDIAN(estimated_occupancy_l365d) AS median_estimated_occupancy,
		ROUND(AVG(estimated_occupancy_l365d), 2) AS avg_estimated_occupancy
	FROM cleaned_listings
	WHERE stay_policy IS NOT NULL AND room_type = 'Entire home/apt' AND bedrooms BETWEEN 1 AND 3 AND stay_policy IN ('Short stay', 'Monthly stay')
	GROUP BY stay_policy, bedrooms
)

SELECT 
	stay_policy,
	bedrooms,
	listings,
	zero_occupancy_listings,
	ROUND(zero_occupancy_listings * 100.0 / listings, 2) AS pct_zero_occupancy
FROM stay_policy_bedrooms_occupancy
ORDER BY stay_policy, bedrooms;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- How does estimated occupancy differ across the major types of Denver Airbnb listings?

SELECT 
	bedrooms,
	COUNT(*) AS listings,
	MEDIAN(estimated_occupancy_l365d) AS median_estimated_occupancy,
	ROUND(AVG(estimated_occupancy_l365d), 2) AS avg_estimated_occupancy
FROM cleaned_listings
WHERE room_type = 'Entire home/apt' AND bedrooms BETWEEN 1 AND 3
GROUP BY bedrooms
ORDER BY avg_estimated_occupancy DESC;
```

## Are amenity-rich listings associated with higher prices, and are there particular amenities associated with price premiums?

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- Are amenity-rich listings associated with higher prices, and are there particular amenities associated with price premiums?

amenity_unnest AS (
    SELECT
        id,
        UNNEST(
            FROM_JSON(amenities, '["VARCHAR"]')
        ) AS amenity
    FROM cleaned_listings
),

has_amenity AS (
	SELECT 
		id,
		MAX(CASE 
			WHEN LOWER(amenity) LIKE '%pool%' AND LOWER(amenity) NOT LIKE '%pool table%' AND LOWER(amenity) NOT LIKE '%pool view%' AND LOWER(amenity) NOT LIKE '%whirlpool%' THEN 1
			ELSE 0
		END) AS has_pool
	FROM amenity_unnest
	GROUP BY id
),

has_amenity_pivot AS (
SELECT 
	l.bedrooms,
	SUM(CASE WHEN a.has_pool = 1 THEN 1 ELSE 0 END) AS pool_listings,
	MEDIAN(CASE WHEN a.has_pool = 1 THEN l.cleaned_price END) AS pool_median_price,
	SUM(CASE WHEN a.has_pool = 0 THEN 1 ELSE 0 END) AS no_pool_listings,
	MEDIAN(CASE WHEN a.has_pool = 0 THEN l.cleaned_price END) AS no_pool_median_price
FROM cleaned_listings AS l
INNER JOIN has_amenity AS a
	ON l.id = a.id
WHERE l.bedrooms <= 3 AND l.room_type = 'Entire home/apt' AND l.cleaned_price IS NOT NULL
GROUP BY l.bedrooms
)

SELECT *,
	ROUND((pool_median_price - no_pool_median_price) * 100.0 / no_pool_median_price, 2) AS pct_pool_premium
FROM has_amenity_pivot
ORDER BY bedrooms;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- Are amenity-rich listings associated with higher prices, and are there particular amenities associated with price premiums?

amenity_comparison AS (
	SELECT 
		bedrooms,
		CASE 
			WHEN LOWER(amenities) LIKE '%hot tub%' THEN 'With hot tub'
			ELSE 'Without hot tub'
		END AS has_amenity,
		COUNT(*) AS listings,
		MEDIAN(cleaned_price) AS median_price
	FROM cleaned_listings 
	WHERE cleaned_price IS NOT NULL AND room_type = 'Entire home/apt' AND bedrooms <= 3
	GROUP BY bedrooms, has_amenity
),

amenity_comparison_pivot AS (
	SELECT 
		bedrooms,
		MAX(CASE WHEN has_amenity = 'With hot tub' THEN listings END) AS hot_tub_listings,
		MAX(CASE WHEN has_amenity = 'Without hot tub' THEN listings END) AS no_hot_tub_listings,
		MAX(CASE WHEN has_amenity = 'With hot tub' THEN median_price END) AS hot_tub_median,
		MAX(CASE WHEN has_amenity = 'Without hot tub' THEN median_price END) AS no_hot_tub_median
	FROM amenity_comparison
	GROUP BY bedrooms
)

SELECT *,
	hot_tub_median - no_hot_tub_median AS price_difference,
	ROUND((hot_tub_median - no_hot_tub_median) * 100.0 / no_hot_tub_median, 2) AS pct_premium
FROM amenity_comparison_pivot
ORDER BY bedrooms;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- Are amenity-rich listings associated with higher prices, and are there particular amenities associated with price premiums?

amenity_unnest AS (
    SELECT
        id,
        UNNEST(
            FROM_JSON(amenities, '["VARCHAR"]')
        ) AS amenity
    FROM cleaned_listings
)

SELECT
    amenity,
    COUNT(DISTINCT id) AS listings
FROM amenity_unnest
WHERE LOWER(amenity) LIKE '%pool%'
GROUP BY amenity
ORDER BY listings DESC;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- Are amenity-rich listings associated with higher prices, and are there particular amenities associated with price premiums?

amenity_unnest AS (
	SELECT
	    id,
		cleaned_price,
	    UNNEST(
	        FROM_JSON(amenities, '["VARCHAR"]')
	    ) AS amenity
	FROM cleaned_listings
)
	
SELECT 
	amenity,
	COUNT(DISTINCT id) AS listings,
	MEDIAN(cleaned_price) AS median_price
FROM amenity_unnest
WHERE amenity IS NOT NULL AND cleaned_price IS NOT NULL
GROUP BY amenity
HAVING COUNT(DISTINCT id) >= 10
ORDER BY median_price DESC;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- Are amenity-rich listings associated with higher prices, and are there particular amenities associated with price premiums?

SELECT 
	bedrooms,
	CASE 
		WHEN amenity_count <= 10 THEN '10 or Less Amenities'
		WHEN amenity_count <= 20 THEN '11-20 Amenities'
		WHEN amenity_count <= 30 THEN '21-30 Amenities'
		WHEN amenity_count <= 40 THEN '31-40 Amenities' 
		WHEN amenity_count <= 50 THEN '41-50 Amenities'
		WHEN amenity_count <= 60 THEN '51-60 Amenities'
		WHEN amenity_count <= 70 THEN '61-70 Amenities'
		WHEN amenity_count <= 80 THEN '71-80 Amenities'
		ELSE 'Over 80 Amenities'
	END AS amenity_count_category,
	COUNT(*) AS listings,
	MEDIAN(cleaned_price) AS median_price
FROM cleaned_listings 
WHERE cleaned_price IS NOT NULL AND room_type = 'Entire home/apt' AND bedrooms <= 3
GROUP BY bedrooms, amenity_count_category
HAVING COUNT(*) >= 20
ORDER BY bedrooms, amenity_count_category;
```

## Within each room type, how does median nightly price differ across stay policies? | Within comparable bedroom counts, how does median nightly price differ between short-stay and monthly-stay listings?

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- Within (Entire home/apt) comparable bedroom counts, how does median nightly price differ between short-stay and monthly-stay listings?

bedrooms_stay_policy AS (
	SELECT 
		bedrooms,
		stay_policy,
		COUNT(*) AS listings,
		MEDIAN(cleaned_price) AS median_price
	FROM cleaned_listings
	WHERE bedrooms <= 3 AND room_type = 'Entire home/apt' AND stay_policy IN ('Short stay', 'Monthly stay') AND cleaned_price IS NOT NULL 
	GROUP BY bedrooms, stay_policy
),

bedrooms_stay_policy_pivot AS (
SELECT 
	bedrooms,
	MAX(CASE WHEN stay_policy = 'Short stay' THEN listings END) AS short_listings,
	MAX(CASE WHEN stay_policy = 'Monthly stay' THEN listings END) AS monthly_listings,
	MAX(CASE WHEN stay_policy = 'Short stay' THEN median_price END) AS short_median,
	MAX(CASE WHEN stay_policy = 'Monthly stay' THEN median_price END) AS monthly_median
FROM bedrooms_stay_policy
GROUP BY bedrooms
)

SELECT *,
	short_median - monthly_median AS median_difference,
	ROUND((short_median - monthly_median) * 100.0 / short_median, 2) AS pct_monthly_discount
FROM bedrooms_stay_policy_pivot
ORDER BY bedrooms;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- How does minimum-night strategy relate to nightly price?

SELECT 
	stay_policy,
	COUNT(*) AS listings,
	ROUND(AVG(minimum_nights), 2) AS avg_minimum_nights,
	ROUND(MEDIAN(cleaned_price), 2) AS median_price
FROM cleaned_listings
WHERE minimum_nights IS NOT NULL AND cleaned_price IS NOT NULL
GROUP BY stay_policy
ORDER BY median_price DESC;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- How does host scale relate to nightly price? Do large/professional operators charge systematically different prices than single-listing and smaller hosts?
-- Among Entire home/apt listings, how does median nightly price vary by host scale?
-- Among Entire home/apt listings with the same bedroom count, how does median price vary by host scale?
-- Within the same neighborhood and bedroom count, do single-listing hosts generally have higher median prices than multi-listing hosts?
-- After controlling for neighborhood and bedroom count, do Single-listing hosts price differently from Large/professional operators?

host_scale_prices AS (
	SELECT
		neighbourhood_cleansed,
		bedrooms,
		host_scale,
		COUNT(DISTINCT host_id) AS distinct_hosts,
		COUNT(*) AS listings,
		MEDIAN(cleaned_price) AS median_price
	FROM cleaned_listings
	WHERE cleaned_price IS NOT NULL 
		AND room_type = 'Entire home/apt'
		AND bedrooms IS NOT NULL AND bedrooms <= 3 
		AND host_scale IN ('Single-listing host', 'Large/professional operator')
	GROUP BY neighbourhood_cleansed, bedrooms, host_scale
),

host_scale_comparison AS (
	SELECT 
		neighbourhood_cleansed,
		bedrooms,
		SUM(CASE WHEN host_scale = 'Single-listing host' THEN listings ELSE 0 END) AS single_listings,
		SUM(CASE WHEN host_scale = 'Large/professional operator' THEN listings ELSE 0 END) AS professional_listings
	FROM host_scale_prices
	GROUP BY neighbourhood_cleansed, bedrooms
),

thresholds AS (
	SELECT *
	FROM (VALUES
		(1),
		(2),
		(3),
		(5),
		(10),
		(15)
	) AS t(minimum_priced_listings)
),

total_listings AS (
	SELECT 
		COUNT(*) AS total_listings
	FROM cleaned_listings
	WHERE cleaned_price IS NOT NULL 
		AND room_type = 'Entire home/apt'
		AND bedrooms IS NOT NULL AND bedrooms <= 3 
		AND host_scale IN ('Single-listing host', 'Large/professional operator')
)

SELECT 
	t.minimum_priced_listings,
	COUNT(*) AS groups_retained,
	SUM(h.single_listings + h.professional_listings) AS listings_retained,
	ROUND(SUM(h.single_listings + h.professional_listings) * 100.0 / l.total_listings, 2) AS percent_of_listings_retained
FROM host_scale_comparison AS h
INNER JOIN thresholds AS t 
	ON h.single_listings >= t.minimum_priced_listings
	AND h.professional_listings >= t.minimum_priced_listings
CROSS JOIN total_listings AS l
GROUP BY t.minimum_priced_listings, l.total_listings, 
ORDER BY t.minimum_priced_listings;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- How does host scale relate to nightly price? Do large/professional operators charge systematically different prices than single-listing and smaller hosts?
-- Among Entire home/apt listings, how does median nightly price vary by host scale?
-- Among Entire home/apt listings with the same bedroom count, how does median price vary by host scale?
-- Within the same neighborhood and bedroom count, do single-listing hosts generally have higher median prices than multi-listing hosts?

host_scale_prices AS (
	SELECT
		neighbourhood_cleansed,
		bedrooms,
		host_scale,
		COUNT(DISTINCT host_id) AS distinct_hosts,
		COUNT(*) AS listings,
		MEDIAN(cleaned_price) AS median_price
	FROM cleaned_listings
	WHERE cleaned_price IS NOT NULL 
		AND room_type = 'Entire home/apt'
		AND bedrooms IS NOT NULL AND bedrooms <= 3 
		AND host_scale IN ('Single-listing host', 'Large/professional operator')
	GROUP BY neighbourhood_cleansed, bedrooms, host_scale
),

host_scale_comparison AS (
	SELECT 
		neighbourhood_cleansed,
		bedrooms,
		SUM(CASE WHEN host_scale = 'Single-listing host' THEN listings ELSE 0 END) AS single_listings,
		SUM(CASE WHEN host_scale = 'Large/professional operator' THEN listings ELSE 0 END) AS professional_listings,
		MAX(CASE WHEN host_scale = 'Single-listing host' THEN median_price END) AS single_median,
		MAX(CASE WHEN host_scale = 'Large/professional operator' THEN median_price END) AS professional_median
	FROM host_scale_prices
	GROUP BY neighbourhood_cleansed, bedrooms
),

host_scale_medians AS (
	SELECT 
		neighbourhood_cleansed,
		bedrooms,
		single_median,	
		professional_median,
		single_median - professional_median AS difference,
		ROUND((single_median - professional_median) * 100.0 / professional_median, 2) AS pct_difference
	FROM host_scale_comparison
	WHERE single_listings >= 3 AND professional_listings >= 3
),

host_comparison_result AS (
	SELECT 
		CASE 
			WHEN difference > 0 THEN 'Single-listing host more expensive'
			WHEN difference < 0 THEN 'Professional operator more expensive'
			WHEN difference = 0 THEN 'Same median price'
		END AS comparison_result,
		COUNT(*) AS groups,
		ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER(), 2) AS pct_of_groups
	FROM host_scale_medians
	GROUP BY comparison_result
)

SELECT 
	MEDIAN(pct_difference) AS median_pct_difference_across_all_groups,
	MEDIAN(CASE WHEN difference > 0 THEN pct_difference END) AS median_pct_difference_when_single_more_expensive,
	MEDIAN(ABS(pct_difference)) AS median_absolute_pct_difference
FROM host_scale_medians;
```

## How does host scale relate to nightly price? Do large/professional operators charge systematically different prices than single-listing and smaller hosts?

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- How does host scale relate to nightly price? Do large/professional operators charge systematically different prices than single-listing and smaller hosts?

host_scale_prices AS (
	SELECT
		host_scale,
		COUNT(DISTINCT host_id) AS distinct_hosts,
		COUNT(*) AS listings,
		MEDIAN(cleaned_price) AS median_price
	FROM cleaned_listings
	WHERE cleaned_price IS NOT NULL 
	GROUP BY host_scale
),

overall_median AS (
	SELECT 
		MEDIAN(cleaned_price) AS overall_median
	FROM cleaned_listings
	WHERE cleaned_price IS NOT NULL 
)

SELECT *,
	ROUND((h.median_price - m.overall_median) * 100.0 / m.overall_median, 2) AS median_percent_change
FROM host_scale_prices AS h
CROSS JOIN overall_median AS m
ORDER BY listings DESC;
```

## Within Entire home/apt listings, how does capacity/property size affect price? | Among similarly sized Entire homes/apartments, which Denver neighborhoods command higher or lower nightly prices? | For 1-, 2-, and 3-bedroom Entire home/apts, how far above or below the Denver-wide median for the same bedroom count is each qualifying neighborhood's median price?

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- Among similarly sized Entire homes/apartments, which Denver neighborhoods command higher or lower nightly prices?
-- For 1-, 2-, and 3-bedroom Entire home/apts, how far above or below the Denver-wide median for the same bedroom count is each qualifying neighborhood's median price?

neighbourhood_bedroom_prices AS (
	SELECT 
		neighbourhood_cleansed,
		bedrooms,
		COUNT(*) AS priced_listings,
		MEDIAN(cleaned_price) AS median_price
	FROM cleaned_listings
	WHERE room_type = 'Entire home/apt' AND cleaned_price IS NOT NULL AND bedrooms IS NOT NULL AND bedrooms <= 3
	GROUP BY neighbourhood_cleansed, bedrooms
	HAVING COUNT(*) >= 5
),

median_prices AS (
	SELECT 
		bedrooms,
		MEDIAN(cleaned_price) AS entirehomeapt_bedrooms_median_price
	FROM cleaned_listings
	WHERE room_type = 'Entire home/apt' AND cleaned_price IS NOT NULL AND bedrooms IS NOT NULL AND bedrooms <= 3
	GROUP BY bedrooms
),
	
neighbourhood_bedroom_prices_rank AS (
	SELECT n.*, 
		m.entirehomeapt_bedrooms_median_price,
		ROUND((n.median_price - m.entirehomeapt_bedrooms_median_price) * 100.0 / m.entirehomeapt_bedrooms_median_price, 2) AS median_percent_change,
		ROW_NUMBER() OVER (
			PARTITION BY n.bedrooms
			ORDER BY n.median_price DESC
		) AS bedroom_median_price_high_rank,
		ROW_NUMBER() OVER (
			PARTITION BY n.bedrooms
			ORDER BY n.median_price
		) AS bedroom_median_price_low_rank
	FROM neighbourhood_bedroom_prices AS n
	INNER JOIN median_prices AS m
		ON n.bedrooms = m.bedrooms
)

SELECT *,
	CASE 
		WHEN bedroom_median_price_high_rank <= 5 THEN 'Top 5 Premium'
		WHEN bedroom_median_price_low_rank <= 5 THEN 'Top 5 Discount'
		WHEN median_percent_change > 0 THEN 'Premium'
		WHEN median_percent_change < 0 THEN 'Discount'
	END AS price_position
FROM neighbourhood_bedroom_prices_rank 
WHERE bedroom_median_price_high_rank <= 5 OR bedroom_median_price_low_rank <= 5
ORDER BY bedrooms DESC, median_price DESC;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- Among similarly sized Entire homes/apartments, which Denver neighborhoods command higher or lower nightly prices?

neighbourhood_bedroom_prices AS (
	SELECT 
		neighbourhood_cleansed,
		bedrooms,
		COUNT(*) AS priced_listings,
		MEDIAN(cleaned_price) AS median_price
	FROM cleaned_listings
	WHERE room_type = 'Entire home/apt' AND cleaned_price IS NOT NULL AND bedrooms IS NOT NULL AND bedrooms <= 3
	GROUP BY neighbourhood_cleansed, bedrooms
),

threshold AS (
	SELECT *
	FROM (VALUES
		(5), (10), (15), (20), (25), (30), (50)
	) AS t(minimum_priced_listings)
),

neighbourhood_bedroom_prices_threshold AS (
	SELECT 
		t.minimum_priced_listings,
		COUNT(DISTINCT neighbourhood_cleansed) AS retained_neighbourhoods,
		SUM(priced_listings) AS retained_listings
	FROM neighbourhood_bedroom_prices AS n
	INNER JOIN threshold AS t ON n.priced_listings >= t.minimum_priced_listings
	GROUP BY t.minimum_priced_listings
),

total_listings AS (
	SELECT 
		COUNT(*) AS total_listings
	FROM cleaned_listings
	WHERE room_type = 'Entire home/apt' AND cleaned_price IS NOT NULL AND bedrooms IS NOT NULL AND bedrooms <= 3
)

SELECT n.*,
	ROUND(n.retained_listings * 100.0 / t.total_listings, 2) AS percent_of_listings
FROM neighbourhood_bedroom_prices_threshold AS n
CROSS JOIN total_listings AS t
ORDER BY n.minimum_priced_listings;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- Within Entire home/apt listings, how does capacity/property size affect price?

SELECT 
	accommodates,
	COUNT(*) AS priced_listings,
	MEDIAN(cleaned_price) AS median_price
FROM cleaned_listings
WHERE room_type = 'Entire home/apt' AND cleaned_price IS NOT NULL 
GROUP BY accommodates
ORDER BY median_price DESC;
```

## Do neighborhood price differences persist when we compare the same room type? | Among neighborhoods with sufficient Entire home/apt inventory, which neighborhoods have the highest and lowest median nightly prices for Entire homes/apartments?

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- Among neighborhoods with sufficient Entire home/apt inventory, which neighborhoods have the highest and lowest median nightly prices for Entire homes/apartments?

neighbourhood_pricing_entirehomeapt AS (
	SELECT 
		neighbourhood_cleansed,
		COUNT(*) AS priced_listings,
		MEDIAN(cleaned_price) AS median_price
	FROM cleaned_listings 
	WHERE cleaned_price IS NOT NULL AND room_type = 'Entire home/apt'
	GROUP BY neighbourhood_cleansed
),

threshold AS (
	SELECT *
	FROM (VALUES
		(5), (10), (15), (20), (25), (30), (50)
	) AS t(minimum_priced_listings)
),

neighbourhood_pricing_entirehomeapt_thresholds AS (
	SELECT 
		t.minimum_priced_listings,
		COUNT(neighbourhood_cleansed) AS neighbourhoods_retained,
		SUM(priced_listings) AS priced_listings_retained
	FROM neighbourhood_pricing_entirehomeapt AS n
	INNER JOIN threshold AS t ON n.priced_listings >= t.minimum_priced_listings
	GROUP BY t.minimum_priced_listings
),

total_entirehomeapt_listings AS (
	SELECT 
		COUNT(*) AS total_entirehomeapt_listings
	FROM cleaned_listings 
	WHERE cleaned_price IS NOT NULL AND room_type = 'Entire home/apt'
)

SELECT n.*,
	ROUND(n.priced_listings_retained * 100.0 / t.total_entirehomeapt_listings, 2) AS percent_of_listings_retained
FROM neighbourhood_pricing_entirehomeapt_thresholds AS n
CROSS JOIN total_entirehomeapt_listings AS t
ORDER BY n.minimum_priced_listings;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- Which Denver neighborhoods are the most and least expensive based on median nightly price?

neighbourhood_pricing AS (
	SELECT
		neighbourhood_cleansed,
		COUNT(*) AS priced_listings,
		MEDIAN(cleaned_price) AS median_price,
		ROUND(AVG(cleaned_price), 2) AS avg_price,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1_price,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3_price
	FROM cleaned_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY neighbourhood_cleansed
),
	
-- How many neighborhoods and priced listings would remain if we required at least 5, 10, 20, 30, and 50 priced listings per neighborhood?

thresholds AS (
	SELECT *
	FROM (VALUES
		(5),
		(10),
		(20),
		(30),
		(50)
	) AS t(minimum_priced_listings)
),

neighbourhood_pricing_thresholds AS (
	SELECT 
		t.minimum_priced_listings,
		COUNT(n.neighbourhood_cleansed) AS neighbourhoods_retained,
		SUM(n.priced_listings) AS priced_listings_retained
	FROM neighbourhood_pricing AS n
	INNER JOIN thresholds as t 
		ON n.priced_listings >= t.minimum_priced_listings
	GROUP BY t.minimum_priced_listings
),

total_priced_listings AS (
	SELECT 
		COUNT(*) AS total_priced_listings
	FROM cleaned_listings
	WHERE cleaned_price IS NOT NULL
)

SELECT n.*,
	ROUND(n.priced_listings_retained * 100.0 / t.total_priced_listings, 2) AS percent_of_priced_listings_retained
FROM neighbourhood_pricing_thresholds AS n
CROSS JOIN total_priced_listings AS t
ORDER BY n.minimum_priced_listings;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- Which Denver neighborhoods are the most and least expensive based on median nightly price?

SELECT
	neighbourhood_cleansed,
	COUNT(*) AS priced_listings,
	MEDIAN(cleaned_price) AS median_price,
	ROUND(AVG(cleaned_price), 2) AS avg_price,
	ROUND(AVG(cleaned_price) - MEDIAN(cleaned_price), 2) AS avg_median_difference,
	QUANTILE_CONT(cleaned_price, 0.25) AS q1_price,
	QUANTILE_CONT(cleaned_price, 0.75) AS q3_price,
	QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
FROM cleaned_listings
WHERE cleaned_price IS NOT NULL
GROUP BY neighbourhood_cleansed
HAVING COUNT(*) >= 20
ORDER BY median_price DESC;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- How sensitive are room-type price estimates to high-price outliers?

SELECT 
	room_type,
	ROUND(AVG(cleaned_price), 2) AS full_data_mean,
	ROUND(AVG(CASE WHEN price_status != 'High price outlier' THEN cleaned_price END), 2) AS typical_range_mean,
	ROUND(AVG(cleaned_price) - AVG(CASE WHEN price_status != 'High price outlier' THEN cleaned_price END), 2) AS change_in_mean,
	ROUND(
		(
		AVG(cleaned_price) 
		- AVG(CASE WHEN price_status != 'High price outlier' THEN cleaned_price END) 
		)
		* 100.0 / AVG(cleaned_price), 
		2) AS pct_change_in_mean,
	MEDIAN(cleaned_price) AS full_data_median,
	MEDIAN(CASE WHEN price_status != 'High price outlier' THEN cleaned_price END) AS typical_range_median
FROM cleaned_listings
WHERE cleaned_price IS NOT NULL 
GROUP BY room_type
ORDER BY room_type;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- How do nightly prices differ by room type?

SELECT 
	room_type,
	COUNT(*) AS priced_listings,
	MEDIAN(cleaned_price) AS median_price,
	ROUND(AVG(cleaned_price), 2) AS average_price,
	ROUND(AVG(cleaned_price) - MEDIAN(cleaned_price), 2) AS avg_median_difference,
	QUANTILE_CONT(cleaned_price, 0.25) AS q1_price,
	QUANTILE_CONT(cleaned_price, 0.75) AS q3_price
FROM cleaned_listings
WHERE cleaned_price IS NOT NULL
GROUP BY room_type
ORDER BY priced_listings DESC;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

-- How much of the Denver market is classified as a high-price outlier, and how does that vary by room type?
	
SELECT 
	room_type,
	price_status,
	COUNT(*) AS listings,
	ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (PARTITION BY room_type), 2) AS percent_of_room_type
FROM cleaned_listings
WHERE price_status != 'Missing price'
GROUP BY room_type, price_status
ORDER BY room_type, listings DESC;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- What does the overall distribution of nightly listing prices look like in Denver, and how strongly is it affected by extreme values?

price_distribution AS (
SELECT 
	COUNT(cleaned_price) AS non_null_priced_listings,
	MIN(cleaned_price) AS min_price,
	QUANTILE_CONT(cleaned_price, 0.25) AS q1_price,
	MEDIAN(cleaned_price) AS median_price,
	ROUND(AVG(cleaned_price), 2) AS avg_price,
	QUANTILE_CONT(cleaned_price, 0.75) AS q3_price,
	MAX(cleaned_price) AS max_price,
	SUM(CASE WHEN cleaned_price > 500 THEN 1 ELSE 0 END) AS price_over_500,
	SUM(CASE WHEN cleaned_price > 1000 THEN 1 ELSE 0 END) AS price_over_1000,
	SUM(CASE WHEN cleaned_price > 2000 THEN 1 ELSE 0 END) AS price_over_2000,
	SUM(CASE WHEN cleaned_price > 5000 THEN 1 ELSE 0 END) AS price_over_5000,
	SUM(CASE WHEN cleaned_price > 10000 THEN 1 ELSE 0 END) AS price_over_10000
FROM cleaned_listings
)

SELECT 
	non_null_priced_listings,
	price_over_500, 
	ROUND(price_over_500 * 100.0 / non_null_priced_listings, 2) AS over_500_pct_of_price_listings,
	price_over_1000,
	ROUND(price_over_1000 * 100.0 / non_null_priced_listings, 2) AS over_1000_pct_of_price_listings,
	price_over_2000,
	ROUND(price_over_2000 * 100.0 / non_null_priced_listings, 2) AS over_2000_pct_of_price_listings,
	price_over_5000,
	ROUND(price_over_5000 * 100.0 / non_null_priced_listings, 2) AS over_5000_pct_of_price_listings,
	price_over_10000,
	ROUND(price_over_10000 * 100.0 / non_null_priced_listings, 2) AS over_10000_pct_of_price_listings
FROM price_distribution;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- How does host scale vary across Denver neighborhoods, and are particular neighborhoods disproportionately represented by multi-listing/professional hosts?

neighbourhood_host_scale AS (
	SELECT 
		neighbourhood_cleansed,
		host_scale,
		COUNT(DISTINCT host_id) AS distinct_hosts,
		COUNT(*) AS listings,
		SUM(COUNT(DISTINCT host_id)) OVER (PARTITION BY neighbourhood_cleansed) AS total_hosts_in_neighbourhood,
		SUM(COUNT(DISTINCT host_id)) OVER (PARTITION BY host_scale) AS total_host_scale_hosts,
		SUM(COUNT(*)) OVER (PARTITION BY neighbourhood_cleansed) AS listings_in_neighbourhood,
		SUM(COUNT(*)) OVER (PARTITION BY host_scale) AS host_scale_listings
	FROM cleaned_listings
	GROUP BY neighbourhood_cleansed, host_scale
),

neighbourhood_host_scale_percentages AS (
	SELECT 
		neighbourhood_cleansed AS "Neighbourhood",
		host_scale AS "Host Scale",
		distinct_hosts AS "Distinct Hosts",
		total_hosts_in_neighbourhood AS "Total Hosts in Neighbourhood",
		ROUND(distinct_hosts * 100.0 / total_hosts_in_neighbourhood, 2) AS "Percent of Neighbourhood Hosts",
		listings AS "Listings",
		ROUND(listings * 100.0 / listings_in_neighbourhood, 2) AS "Percent of Neighbourhood Listings"
	FROM neighbourhood_host_scale
),

neighbourhood_host_scale_rank AS (
	SELECT 
		"Neighbourhood",
		"Host Scale",
		"Distinct Hosts",
		"Total Hosts in Neighbourhood",
		"Percent of Neighbourhood Hosts",
		ROW_NUMBER() OVER (PARTITION BY "Neighbourhood" ORDER BY "Percent of Neighbourhood Hosts" DESC) AS "Neighbourhood Hosts Percentage Rank",
		"Listings",
		"Percent of Neighbourhood Listings",
		ROW_NUMBER() OVER (PARTITION BY "Neighbourhood" ORDER BY "Percent of Neighbourhood Listings" DESC) AS "Neighbourhood Listings Percentage Rank"
	FROM neighbourhood_host_scale_percentages
)

SELECT * 
FROM neighbourhood_host_scale_rank
WHERE "Neighbourhood Listings Percentage Rank" = 1
ORDER BY neighbourhood;
```

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- How does host scale vary across Denver neighborhoods, and are particular neighborhoods disproportionately represented by multi-listing/professional hosts?

neighbourhood_host_scale AS (
	SELECT 
		neighbourhood_cleansed,
		host_scale,
		COUNT(DISTINCT host_id) AS distinct_hosts,
		COUNT(*) AS listings,
		SUM(COUNT(DISTINCT host_id)) OVER (PARTITION BY neighbourhood_cleansed) AS total_hosts_in_neighbourhood,
		SUM(COUNT(DISTINCT host_id)) OVER (PARTITION BY host_scale) AS total_host_scale_hosts,
		SUM(COUNT(*)) OVER (PARTITION BY neighbourhood_cleansed) AS listings_in_neighbourhood,
		SUM(COUNT(*)) OVER (PARTITION BY host_scale) AS host_scale_listings
	FROM cleaned_listings
	GROUP BY neighbourhood_cleansed, host_scale
)

SELECT 
	neighbourhood_cleansed AS "Neighbourhood",
	host_scale AS "Host Scale",
	distinct_hosts AS "Distinct Hosts",
	ROUND(distinct_hosts * 100.0 / total_hosts_in_neighbourhood, 2) AS "Percent of Neighbourhood Hosts",
	listings AS "Listings",
	ROUND(listings * 100.0 / listings_in_neighbourhood, 2) AS "Percent of Neighbourhood Listings"
FROM neighbourhood_host_scale
WHERE host_scale = 'Large/professional operator'
	AND total_hosts_in_neighbourhood >= 5
	AND listings_in_neighbourhood >= 10
ORDER BY "Percent of Neighbourhood Listings" DESC, "Percent of Neighbourhood Hosts" DESC;
```

## Which neighborhoods have unusually high concentrations of each room type?

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- Which neighborhoods have unusually high concentrations of each room type?

neighbourhood_listings AS (
	SELECT 
		n.neighbourhood,
		l.room_type,
		COUNT(l.id) AS listings,
		SUM(COUNT(l.id)) OVER (PARTITION BY n.neighbourhood) AS neighbourhood_total,
		SUM(COUNT(l.id)) OVER (PARTITION BY l.room_type) AS room_type_total,
		SUM(COUNT(l.id)) OVER () AS denver_total
	FROM 'neighbourhoods.csv' AS n
	LEFT JOIN cleaned_listings AS l
		ON n.neighbourhood = l.neighbourhood_cleansed
	GROUP BY n.neighbourhood, l.room_type
),

neighbourhood_listings_percentages AS (
	SELECT 
		neighbourhood AS "Neighbourhood",
		room_type AS "Room Type",
		listings AS "Neighbourhood Listings",
		neighbourhood_total AS "Neighbourhood Total",
		ROUND(listings * 100.0 / NULLIF(neighbourhood_total, 0), 2) AS "Neighbourhood Room Type Percentage",
		room_type_total AS "Room Type Total",
		ROUND(listings/room_type_total * 100, 2) AS "Percent of Denver Room Type Listings",
		ROUND(room_type_total * 100.0 / denver_total, 2) AS "Denver Room Type Share",
		ROUND(
			(listings * 1.0 / NULLIF(neighbourhood_total, 0))
			/
			(room_type_total * 1.0 / denver_total),
		2
		) AS "Location Quotient (LQ)"
	FROM neighbourhood_listings
	WHERE room_type IS NOT NULL
		AND listings >= 5
)

SELECT *,
	ROW_NUMBER() OVER (
		PARTITION BY "Room Type" 
		ORDER BY "Location Quotient (LQ)" DESC, "Neighbourhood Listings" DESC
	) AS "LQ Rank by Room Type"
FROM neighbourhood_listings_percentages 
WHERE "Location Quotient (LQ)" > 1
ORDER BY "LQ Rank by Room Type", "Location Quotient (LQ)" DESC;
```

## How many listings and what percentage of Denver's listings belong to each room_type?

```python
WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
),

-- How many listings and what percentage of Denver's listings belong to each room_type? 
	
room_type_listings AS (
	SELECT 
		room_type,
		COUNT(*) AS listings
	FROM cleaned_listings
	GROUP BY room_type
),
	
total AS (
	SELECT 
		COUNT(*) AS total_listings
	FROM cleaned_listings
)

SELECT 
	r.room_type AS "Room Type",
	r.listings AS "Listings",
	ROUND(r.listings * 100.0 / t.total_listings, 2) AS "Percent of All Listings",
	ROUND(SUM(r.listings) OVER (ORDER BY r.listings DESC) * 100.0 / SUM(r.listings) OVER (), 2) AS "Cumulative Percentage"
FROM room_type_listings AS r
CROSS JOIN total AS t
ORDER BY "Listings" DESC;
```

# Official Denver Airbnb Data Analysis-Ready CTEs 

```python
/* Official Denver Airbnb Data Analysis-Ready CTEs

- cleaned_listings v1 (frozen on Oct 1, 2026)
- cleaned_calendar v1 (frozen on Oct 1, 2026)
- cleaned_reviews v1 (frozen on Oct 1, 2026)

neighbourhoods.csv (keeping as a reference table)
- no cleaning CTE needed
- 1 neighbourhood is present in this dataset while not being present in the cleaned_listings snapshot - Sun Valley
- cleaned_listings.neighbourhood_cleansed serves as the primary geographic reference 
- neighbourhoods.csv is retained only as a reference for analyses requiring all defined neighborhoods, including those with zero listings.

*/

-- cleaned_listings

WITH base_listings AS (
	SELECT 
		id, last_scraped, source, name,
	
		host_id, host_name, host_location, 
		CASE 
			WHEN host_location IS NULL THEN 'No Location'
			WHEN LOWER(host_location) LIKE '%denver%' THEN 'Denver'
			WHEN TRIM(LOWER(host_location)) LIKE '%, co' OR LOWER(host_location) LIKE '%colorado%' THEN 'Other Colorado'
			WHEN TRIM(LOWER(host_location)) LIKE '%, __' OR LOWER(host_location) LIKE '%united states%' THEN 'Other US State'
			ELSE 'Other Country'
		END AS host_location_category, -- broader category for host_location 
	
		host_is_superhost, host_listings_count, calculated_host_listings_count, host_identity_verified, hosts_time_as_host_years,
		CASE
			WHEN calculated_host_listings_count = 1 THEN 'Single-listing host'
			WHEN calculated_host_listings_count BETWEEN 2 AND 4 THEN 'Small multi-listing host'
			WHEN calculated_host_listings_count BETWEEN 5 AND 9 THEN 'Medium operator'
			WHEN calculated_host_listings_count >= 10 THEN 'Large/professional operator'
		END AS host_scale, -- category column for calculated_host_listings_count ranges
	
		neighbourhood_cleansed, latitude, longitude,
	
		property_type, 
		CASE 
			WHEN property_type = 'Tiny home' THEN 'Specialty/Other'
			WHEN LOWER(property_type) LIKE '%home%' OR LOWER(property_type) LIKE '%bungalow%' OR LOWER(property_type) LIKE '%cottage%' THEN 'House/Home'
			WHEN LOWER(property_type) LIKE '%apartment%' OR LOWER(property_type) LIKE '%rental unit%' OR LOWER(property_type) LIKE '%loft%' THEN 'Apartment/Rental Unit'
			WHEN LOWER(property_type) LIKE '%condo%' THEN 'Condo'
			WHEN LOWER(property_type) LIKE '%townhouse%' THEN 'Townhouse'
			WHEN LOWER(property_type) LIKE '%guest%' THEN 'Guesthouse'
			WHEN LOWER(property_type) LIKE '%hotel%' OR LOWER(property_type) LIKE '%hostel%' OR LOWER(property_type) LIKE '%bed and breakfast%' THEN 'Hotel/Hospitality'
			WHEN LOWER(property_type) LIKE '%villa%' THEN 'Villa'
			ELSE 'Speciality/Other'
		END AS property_category, -- broader category for property_type
	
		room_type, accommodates, bedrooms, beds, amenities,
		 LEN(
			FROM_JSON(amenities, '["VARCHAR"]')
    	) AS amenity_count, -- converting amenities column to a list of VARCHAR values and counting amenities in the list for each listing

		bathrooms, bathrooms_text, 
		CASE 
			WHEN LOWER(bathrooms_text) LIKE '%private%' THEN 'Private'
			WHEN LOWER(bathrooms_text) LIKE '%shared%' THEN 'Shared'
			WHEN bathrooms_text IS NOT NULL THEN 'Standard'
			ELSE NULL
		END AS bathroom_type, -- categorizing the type of bathroom from bathrooms_text
	
		price,
		TRY_CAST(
            REPLACE(
                REPLACE(price, '$', ''),
                ',', ''
            ) AS DECIMAL(10,2)
        ) AS cleaned_price, -- converting VARCHAR price values to a usable numeric data type for calculations 
	
		minimum_nights, maximum_nights, has_availability, availability_30, availability_60, availability_90, availability_365, availability_eoy, estimated_occupancy_l365d,  estimated_revenue_l365d,
		CASE 
			WHEN minimum_nights <= 6 THEN 'Short stay'
			WHEN minimum_nights <= 27 THEN 'Medium Stay'
			WHEN minimum_nights <= 45 THEN 'Monthly stay'
			WHEN minimum_nights >= 46 THEN 'Extended stay'
		ELSE NULL
		END AS stay_policy, -- category column for minimum_nights ranges 
	
		first_review, last_review, number_of_reviews, number_of_reviews_ltm, number_of_reviews_l30d, number_of_reviews_ly,
		CASE 
			WHEN number_of_reviews_l30d >= 1 THEN 'Recently Reviewed'
			WHEN number_of_reviews_ltm >= 1 THEN 'Reviewed in Last Year'
			WHEN number_of_reviews >= 1 THEN 'Historically Reviewed'
			WHEN number_of_reviews = 0 THEN 'Never Reviewed'
		END AS listing_activity_status, -- category column for number_of_reviews_... fields
	
		review_scores_rating, review_scores_accuracy, review_scores_communication, review_scores_location, review_scores_value, reviews_per_month
FROM 'detailed_listings.csv'
),
	
room_type_price_quartiles AS (
	SELECT 
		room_type,
		QUANTILE_CONT(cleaned_price, 0.25) AS q1,
		QUANTILE_CONT(cleaned_price, 0.75) AS q3,
		QUANTILE_CONT(cleaned_price, 0.75) - QUANTILE_CONT(cleaned_price, 0.25) AS iqr
	FROM base_listings
	WHERE cleaned_price IS NOT NULL
	GROUP BY room_type
), -- intermediate CTE to calculate respective q1, q3, and iqr for each room_type

room_type_price_stats AS (
	SELECT 
		room_type,
		q1,
		q3,
		iqr,
		q1 - (1.5 * iqr) AS lower_bound,
		q3 + (1.5 * iqr) AS upper_bound
	FROM room_type_price_quartiles
), -- intermediate CTE to consolidate quartile values and calculated lower and upper bounds for each room_type

cleaned_listings AS (
	SELECT 
		b.*,
		CASE 
			WHEN b.cleaned_price IS NULL THEN 'Missing price'
			WHEN b.cleaned_price < p.lower_bound THEN 'Low price outlier'
			WHEN b.cleaned_price > p.upper_bound THEN 'High price outlier'
			ELSE 'Typical range' 
		END AS price_status -- category column for price, based on respective calculations for each room_type
	FROM base_listings AS b
	LEFT JOIN room_type_price_stats AS p
		ON b.room_type = p.room_type
), -- final cleaned_listings CTE

-- cleaned_calendar
	
cleaned_calendar AS (
	SELECT 
		c.listing_id,	
		CAST(c.date AS DATE) AS date,
		c.available,
		c.minimum_nights,
		c.maximum_nights,
		YEAR(c.date) AS year,
		MONTH(c.date) AS month,
		CAST(DATE_TRUNC('month', c.date) AS DATE) AS year_month
	FROM 'detailed_calendar.csv' AS c
	INNER JOIN cleaned_listings AS l
		ON c.listing_id = l.id
),

-- cleaned_reviews 

cleaned_reviews AS (
	SELECT 
		r.listing_id,
		r.id AS review_id,
		CAST(r.date AS DATE) AS date,
		r.reviewer_id,
		r.reviewer_name,
		r.comments,
		YEAR(r.date) AS year,
		MONTH(r.date) AS month,
		CAST(DATE_TRUNC('month', r.date) AS DATE) AS year_month
	FROM 'detailed_reviews.csv' AS r
	INNER JOIN cleaned_listings AS l
		ON r.listing_id = l.id
)

SELECT *
FROM cleaned_listings;
```

- Fourth, location is also somewhat lower than operational categories such as communication and check-in. That's interesting because location is largely a property characteristic rather than something a host can improve operationally.
- Listings with zero reviews have NULL review scores, while listings with at least one review have values for all seven review-score fields.
- Review-score fields contain 638 NULL values, corresponding exactly to listings with zero reviews. These NULLs represent unrated listings rather than erroneous missing data and will therefore be retained.

----------

host 
- Single-listing hosts make up 83.91% of all hosts, but account for only 56.37% of Denver's Airbnb listings.
- Hosts with more than 10 listings represent only 0.96% of hosts, yet control 15.02% of the listings in this dataset.
- 83.91% of hosts have exactly one listing, accounting for 56.37% of listings.
- 16.09% of hosts have multiple listings, accounting for 43.63% of listings.
- Hosts with more than 10 listings are only 0.96% of hosts, but account for 15.02% of listings.
- The 10 largest individual hosts account for 7.41% of all listings.
- Pricing: Superhost listings have a median asking price of $206 versus $135.82, roughly 52% higher. The average is also substantially higher ($323.77 vs. $212.46). Because both groups have averages substantially above their medians, that's another indication that price is right-skewed and reinforces our decision to pay attention to medians.
- Review activity: Superhost listings average 2.09 reviews/month versus 1.15 for non-Superhosts. That's a more useful comparison than the raw total-review numbers because there are different numbers of listings in each group.
- Large-host presence varies substantially across Denver neighborhoods. Some neighborhoods have relatively high shares of listings operated by hosts with portfolios exceeding 10 Denver listings.
- Validated calculated_host_listings_count against an independently calculated COUNT(*) by host_id. No discrepancies were identified among the 4,939 Denver listings, indicating that the field accurately represents the number of listings associated with each host within this dataset.

--------

estimated occupancy and revenue 
- Estimated Occupancy: Complete across all 4,939 listings, ranging from 0–255 estimated occupied nights. The median is 70 nights versus a mean of 103.91, indicating a right-skewed distribution. 1,365 listings (27.6%) have zero estimated occupancy. No negative or >365 values were identified.
- Estimated Revenue: Available for 4,251 listings and missing for 688. Median estimated annual revenue is $15,979 compared with a mean of $28,047.32, indicating substantial right skew from high-revenue listings. The maximum is $3,795,253.
- Revenue Nulls: All 688 missing estimated-revenue values correspond exactly to listings with missing prices. No listing with a populated price has missing estimated revenue.
- Metric Dependency: For all comparable records, estimated_revenue_l365d equals price × estimated_occupancy_l365d. Estimated revenue should therefore be treated as a derived estimate, not an independently observed measure of actual Airbnb revenue.

## detailed_calendar.csv

- Grain: one listing-calendar-date per row
- Composite key: (listing_id, date)
- detailed_calendar.csv grain and key: The dataset contains 1,806,385 rows representing 4,949 distinct listings. Each listing has exactly 365 records corresponding to 365 distinct dates. No duplicate (listing_id, date) combinations or NULL values in either key field were identified. Therefore, one row represents one listing on one calendar date, and (listing_id, date) is validated as the composite primary key.
- Cross-dataset integrity: detailed_calendar.csv contains 4,949 listing IDs compared with 4,939 listings in detailed_listings.csv, indicating 10 calendar listing IDs are absent from the current listings snapshot.
- 4,305 listings (87.0%) cover June 30, 2026 → June 29, 2027.
- 644 listings (13.0%) cover July 2, 2026 → July 1, 2027.
- Calendar date coverage: Each of the 4,949 listings contains exactly 365 unique calendar dates, but listings do not all share the same calendar window. 4,305 listings (87.0%) span June 30, 2026 through June 29, 2027, while 644 listings (13.0%) span July 2, 2026 through July 1, 2027. This suggests the calendar records were captured across different scrape dates/batches. Date-level analyses should account for differences in listing coverage near the boundaries of the dataset.

----------

availability
- Availability: The available field is complete and contains only the expected Boolean values. Across all 1,806,385 listing-date records, 1,021,669 (56.56%) are marked available and 784,716 (43.44%) are marked unavailable.
- False means the date is not available on the Airbnb calendar, but we don't know whether that's because it was booked, host-blocked, restricted, or unavailable for some other reason.
- Listing-level calendar availability: Availability varies substantially across Denver listings. The median listing is marked available for 236 of its 365 calendar dates (64.7%), compared with a mean of 206.44 days (56.6%). Approximately 40.9% of listings are available for at least 271 days, while 7.7% have zero available dates. These differences demonstrate substantial variation in calendar availability across listings. Because unavailable dates may represent either reservations or host-imposed restrictions, these values should not be interpreted directly as occupancy.
- Cross-dataset availability validation: Calendar-derived annual availability was compared with detailed_listings.availability_365 for all 4,939 listings represented in both datasets. 4,938 listings (99.98%) matched exactly. One listing was inconsistent: availability_365 reported 0 available days, while its 365 calendar records contained 207 available and 158 unavailable dates. The calendar record itself contained a complete 365-day window and plausible variation in availability over time. Because the source of the discrepancy cannot be determined from the available data, the record is retained and documented as a cross-dataset inconsistency rather than corrected.
- Calendar availability is substantially lower in the immediate future, rises considerably farther into the booking window, peaks around November–December in this snapshot, and then declines somewhat in later spring/summer months.

------------

minimum and maximum nights
- Minimum-stay requirements are considerably more dynamic than maximum-stay requirements. Approximately 24.5% of listings have multiple minimum-night requirements across their 365-day calendar, compared with only about 2.5% having multiple maximum-night requirements.
- That makes intuitive analytical sense as something worth investigating: hosts may configure different minimum stays for different dates. We don't need to determine why yet.
- Calendar stay requirements: minimum_nights and maximum_nights are complete across all 1,806,385 calendar records. Calendar-level minimum stays range from 1–500 nights, while maximum stays range from 1–1,125 nights. Stay requirements are not necessarily constant within a listing's calendar. Of 4,949 listings, 1,213 (24.5%) have multiple minimum-night requirements across their 365 dates, while 125 (2.5%) have multiple maximum-night requirements. Minimum-stay restrictions therefore exhibit substantially greater date-level variation than maximum-stay restrictions.
- The listing-level stay-restriction values are highly consistent with calendar-level restrictions, but calendar data contains additional date-specific variation that the listing-level fields do not capture.
- - Cross-dataset stay-restriction validation: Listing-level minimum_nights and maximum_nights were compared against all calendar-level values for each listing. Among 4,939 listings represented in both datasets, 4,938 (99.98%) had a listing-level minimum-night value occurring on at least one calendar date, and 4,938 (99.98%) had a listing-level maximum-night value occurring on at least one calendar date. Calendar-level restrictions can vary by date, particularly minimum stays, so listing-level fields should not be interpreted as necessarily applying uniformly across the entire 365-day calendar.
- Cross-dataset anomaly: One listing was identified as the sole exception across multiple listing-to-calendar validation tests. Its listing-level record reports minimum_nights = 0, maximum_nights = 0, and availability_365 = 0, while its calendar contains a complete 365-day window with minimum_nights = 1, maximum_nights = 365, and 207 available dates. The remaining 4,938 listings shared by both datasets reconcile successfully across these checks. Because the cause of the discrepancy cannot be established from the source data, the listing should be flagged as an anomalous cross-dataset record rather than automatically corrected.
- Stay-restriction consistency: 3,608 of 1,806,385 calendar records (~0.20%) have minimum_nights > maximum_nights. These records are highly concentrated, affecting only 14 of 4,949 listings (~0.28%). The condition appears logically inconsistent under a straightforward interpretation of minimum- and maximum-stay restrictions, but the records are not automatically classified as source errors because the underlying Airbnb rule semantics may be more complex. The affected listings were flagged for further inspection rather than corrected or removed.
- Calendar stay-rule consistency: 3,608 calendar records (~0.20%) have minimum_nights > maximum_nights, affecting 14 of 4,949 calendar listings (~0.28%). The condition is highly concentrated: 11 listings account for nearly all affected records, with 270–365 affected dates each, while three listings are affected on only one date. Most heavily affected listings have a constant 29-night minimum while their maximum-night restriction varies by date. Because the underlying Airbnb booking-rule semantics cannot be fully determined from the source data, these records are retained and flagged rather than automatically classified as errors or corrected.

## detailed_reviews.csv

- detailed_reviews grain/key: One row represents one individual Airbnb review. id is a valid primary-key candidate in the current dataset: all 354,741 review IDs are non-NULL and unique.
- Reviews/listings relationship: detailed_reviews.listing_id relates reviews to detailed_listings.id. Of 354,741 review records, 95 historical reviews reference 8 listing IDs that are absent from the current listings snapshot. Conversely, 638 current listings have no records in the detailed reviews table. The unmatched historical reviews are retained because the reviews dataset spans many years and may include listings no longer represented in the current snapshot.
- Review-count reconciliation: For all 4,939 current listings, detailed_listings.number_of_reviews exactly equals the number of corresponding records in detailed_reviews.csv. This includes the 638 listings reporting zero lifetime reviews, confirming that their absence from detailed_reviews is consistent with the listing-level review count rather than apparent missing review data.
- Reviewer-field completeness: All 354,741 review records contain both a reviewer_id and reviewer_name. No records contain a reviewer ID without a name or a name without an ID. Reviewer-level analysis should use reviewer_id rather than reviewer_name as the identifier.
- For every listing included by the INNER JOIN, the listing table's first_review and last_review fields exactly agree with the minimum and maximum dates observed in detailed_reviews.csv.
